```python
# === portable-setup (bban4040) ===
# Secrets resolve from Colab "Secrets" (userdata) on Colab, or environment
# variables / a local .env file when running locally. Nothing is hardcoded.
import os
try:
    from dotenv import load_dotenv
    load_dotenv()
except Exception:
    pass


def get_secret(name, default=None):
    try:
        from google.colab import userdata
        value = userdata.get(name)
        if value:
            return value.strip()
    except Exception:
        pass
    value = os.environ.get(name, default)
    return value.strip() if isinstance(value, str) else value


_key = get_secret("OPENAI_API_KEY")
if _key:
    os.environ["OPENAI_API_KEY"] = _key
```

```python
from openai import OpenAI
import panel as pn

client = OpenAI(api_key=get_secret("OPENAI_API_KEY"))
#model = "gpt-3.5-turbo"
model = "gpt-4o-mini"
```

```python
#This function will receive the different messages in the conversation,
#and call OpenAI passing the full conversation.
def continue_conversation(messages, temperature=0):
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
    )
    return response.choices[0].message.content
```

```python
def add_prompts_conversation(_):
    #Get the value introduced by the user
    prompt = client_prompt.value_input
    client_prompt.value = ''
    if not prompt:               # ignore empty clicks (no API call, no welcome spam)
        return pn.Column(*panels)

    #Append to the context the User prompt.
    context.append({'role':'user', 'content':f"{prompt}"})

    #Get the response.
    response = continue_conversation(context)

    #Add the response to the context.
    context.append({'role':'assistant', 'content':f"{response}"})

    #Update the panels to show the conversation.
    panels.append(
        pn.Row('User:', pn.pane.Markdown(prompt, width=600)))
    panels.append(
        pn.Row('Assistant:', pn.pane.Markdown(response, width=600)))

    return pn.Column(*panels)
```

```python
#Creating the system part of the prompt
#Read and understand it.

context = [ {'role':'system', 'content':"""
You work collecting orders in a delivery IceCream shop called
I'm freezed.

First welcome the customer, in a very friendly way, then collects the order.

Your instructions are:
-Collect the entire order, only from options in our menu, toppings included.
-Summarize it
-check for a final time if everything is ok or the customer wants to add anything else.
-collect the payment, be sure to include toppings and the size of the ice cream.
-Make sure to clarify all options, extras and sizes to uniquely
identify the item from the menu.
-Your answer should be short in a very friendly style.

Our Menu:
The IceCream menu includes only the flavors:
-Vanilla.
-Chocolate.
-Lemon.
-Strawberry.
-Coffee.

The IceCreams are available in two sizes:
-Big: 3$
-Medium: 2$

Toppings:
-Caramel sauce
-White chocolate
-melted peanut butter
Each topping cost 0.5$

"""} ]

#Creating the panel.
import warnings
from bokeh.util.warnings import BokehUserWarning

# Colab can keep old Bokeh references when a notebook cell is rerun.
warnings.filterwarnings(
    "ignore",
    message="reference already known.*",
    category=BokehUserWarning,
)
pn.extension()

panels = []

# Greet the customer once, before any input, so the chat opens with a welcome.
greeting = continue_conversation(context)
context.append({'role': 'assistant', 'content': greeting})
panels.append(pn.Row('Assistant:', pn.pane.Markdown(greeting, width=600)))

client_prompt = pn.widgets.TextInput(value="Hi", placeholder='Enter text here…')
button_conversation = pn.widgets.Button(name="talk")

interactive_conversation = pn.bind(add_prompts_conversation, button_conversation)

dashboard = pn.Column(
    client_prompt,
    pn.Row(button_conversation),
    pn.panel(interactive_conversation, loading_indicator=True),
)

#To talk with the chat push the button: 'talk' after your sentence.
dashboard
```
